```python
commodity = pd.read_csv("/Users/atienojacinta/PYDATA /Visualizations/commodity_futures (1).csv")
```

```python
commodity.head()
```

```python
commodity.info()
```

```python
commodity['Date'] = pd.to_datetime(commodity['Date'])
```

```python
commodity['Date'].dtype
```

```python
fig, ax = plt.subplots(2,2,figsize=(16,8))
ax[0,0].plot(commodity['Date'],commodity['GOLD'])
ax[0,1].plot(commodity['Date'],commodity['ALUMINIUM'])
ax[1,0].plot(commodity['Date'],commodity['CORN'])
ax[1,1].plot(commodity['Date'],commodity['COPPER'])
plt.show()
```

```python
plt.figure(figsize=(12,6))
x = commodity['Date']
y = commodity['GOLD']
plt.plot(x,y)
plt.title('Gold Prices overtime',fontdict={'fontsize':25,'fontweight':'medium'})
plt.show()
```

```python
plt.figure(figsize=(12,4))
x = commodity['Date']
y = commodity['GOLD']
z = commodity['ALUMINIUM']
plt.plot(x,y,label = 'Gold',color='green')
plt.plot(x,z,label = 'Aluminium',color='maroon')
plt.title('Gold Prices overtime',fontdict={'fontsize':12,'fontweight':'bold'})
plt.xlabel('Date')
plt.ylabel('Commodity Price')
plt.legend()
plt.show()
```

```python
fig,ax = plt.subplots(1,2,figsize=(12,4))
x = commodity['Date']
y = commodity['GOLD']
z = commodity['ALUMINIUM']
ax[0].plot(x,y,color='beige',label='GOLD')
ax[1].plot(x,z,color='maroon',label = 'ALUMINIUM')
ax[0].legend()
ax[1].legend()
ax[0].set_title('Gold over time')
ax[1].set_title('Aluminium over time')
ax[0].set_xlabel('Date')
ax[1].set_xlabel('Date')
ax[0].set_ylabel('Gold Prices')
ax[1].set_ylabel('Aluminium Prices')
plt.show()
```
