```python
sales_quantium = quantium.groupby('DATE')[['TOT_SALES']].sum().reset_index()
sales_quantium.head()
```

```python
plt.figure(figsize=(12,5))
x = sales_quantium['DATE']
y = sales_quantium['TOT_SALES']
plt.plot(x,y)
plt.title('Sales Overtime')
plt.show()
```

```python
transactions_quantium = quantium.groupby('DATE')[['TOT_SALES']].count().reset_index()
transactions_quantium.head()
```

```python
plt.figure(figsize=(12,5))
a = transactions_quantium['DATE']
b = transactions_quantium['TOT_SALES']
plt.plot(a,b)
plt.title('Sales Overtime')
plt.show()
```

```python
fig, ax = plt.subplots(2,1,figsize=(12,7))
ax[0].plot(x,y)
ax[1].plot(a,b)
plt.show()
```
